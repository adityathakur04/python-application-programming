```python
def zip_two(a, b):
    out = []
    for i in range(max(len(a), len(b))):
        if i < len(a):
            out.append(a[i])
        if i < len(b):
            out.append(b[i])
    return "".join(out)


def zipper_file(filename, strings):
    with open(filename, "w") as f:
        for i in range(0, len(strings), 2):
            f.write(zip_two(strings[i], strings[i + 1]) + "\n")
```

```python
zipper_file("zipper_out.txt", ["abc", "123", "xy", "WXYZ"])
print(open("zipper_out.txt").read())
```

```python
MONTHS = ["January", "February", "March", "April", "May", "June", "July",
          "August", "September", "October", "November", "December"]


def reformat_date(dates):
    result = []
    for d in dates:
        mm, dd, yyyy = d.split("/")
        result.append(f"Office Hours on {MONTHS[int(mm) - 1]} {dd}, {yyyy}")
    return result
```

```python
reformat_date(["01/05/2024", "12/31/2023", "07/04/1999"])
```
