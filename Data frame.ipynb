```python

data = {
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Aman", "Riya", "Rahul", "Priya", "Karan",
             "Neha", "Arjun", "Simran", "Rohit", "Anjali"],
    "Age": [25, 32, 41, 29, 35, 22, 45, 31, 27, 38],
    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female"],
    "Purchase_Amount": [2000, 3500, 1500, 4000, 2500,
                        1800, 5000, 3200, 2200, 4500]
}

customers = pd.DataFrame(data)

print("First 5 rows:")
print(customers.head())

print("\nSummary Statistics:")
print(customers.describe())

print("\nMissing Values:")
print(customers.isnull().sum())

# Handle missing values
customers = customers.ffill()

print("\nDataset after handling missing values:")
print(customers)
```

```python


customers["Discounted_Price"] = (
    customers["Purchase_Amount"] * 0.90
)

print("Dataset with Discounted Price:")
print(customers)


print("\nAverage Purchase Amount by Gender:")
print(
    customers.groupby("Gender")["Purchase_Amount"].mean()
)

sorted_customers = customers.sort_values(
    by="Purchase_Amount",
    ascending=False
)

print("\nCustomers Sorted by Purchase Amount:")
print(sorted_customers)
```
